```python
import pandas as pd# PHASE 3 TASK1 "GMW"
orders = pd.read_csv("datasets/orders.csv")

gmw = orders["final_amount"].sum()

task_1 = print("Gross Merchandise Value :\n",gmw)
```

```python
net_rev = orders[orders["status"] == "Delivered"]["final_amount"].sum()

task_2 = print("Net revenue: \n",net_rev)
```

```python
net_rev = orders[orders["status"] == "Delivered"]["final_amount"].sum()

net_rev

total_delivered_orders = orders[orders["status"] == "Delivered"].count()

total_delivered_orders

aov = net_rev/total_delivered_orders

task_3 = print("Average Order Value :\n",aov)
```

```python
total_orders = orders["order_id"].count()

total_orders

total_cancelled_orders = orders[orders["status"] == "Cancelled"]["order_id"].count()

cancellation_rate = (total_cancelled_orders/total_orders)*100

task_4 = print("Cancellation Rate :\n",cancellation_rate)
```

```python

returns = orders[orders["status"] == "Returned"]["order_id"].count()

return_rate = (returns/total_delivered_orders)*100

task_5= print("Return Rate :\n",return_rate)
```

```python
customer = pd.read_csv("datasets/customers.csv")
```

```python
cust_ord_merged = pd.merge(customer,orders,on="customer_id")

pd.set_option("display.max_columns",None)

cust_group = cust_ord_merged.groupby("customer_id")["final_amount"].sum()

cust_group = pd.merge(
    cust_group,
    customer[["customer_id", "segment"]],
    on="customer_id"
)

clv = cust_group.groupby("segment")["final_amount"].mean()

task_6 = print("Customer Lifetime Value \n", clv)
```

```python
orders = pd.read_csv("datasets/orders.csv")

orders["order_date"] = pd.to_datetime(orders["order_date"],format="%d-%m-%Y")


orders["Month"] = orders["order_date"].dt.to_period("M")

month_wise_final_amnt = orders.groupby("Month")["final_amount"].sum()

month_wise_final_amnt

prev_gmv = month_wise_final_amnt.shift()

growth = month_wise_final_amnt - prev_gmv

growth_percentage = (growth/prev_gmv)*100

task_7 = print("MOM growth\n",growth_percentage)
```

```python
order_items = pd.read_csv("datasets/order_items.csv")

products = pd.read_csv("datasets/products.csv")

order_items

prod_oritems_merge = pd.merge(order_items,products, on="product_id")

prod_oritems_merge

o_p_ot_merge = pd.merge(prod_oritems_merge,orders, on="order_id")

top_category_revenue = o_p_ot_merge.groupby("category_x")["final_amount"].sum().sort_values(ascending=False)

task_8 = print("top category revenue \n\n",top_category_revenue)
```

```python
payments = pd.read_csv("datasets/payments.csv")

total_payments = payments["payment_id"].count()

failed_payments = payments[payments["status"] == "Failed"]["payment_id"].count()

payment_failure_rate = (failed_payments/total_payments)*100

task_9 = print("payment failure rate :",payment_failure_rate)
```

```python
inventory = pd.read_csv("datasets/inventory.csv")

in_stock_sku = len(inventory[inventory['quantity_available']>0])

total_sku = len(inventory)

inventory_fill_rate = (in_stock_sku/total_sku)*100
```
